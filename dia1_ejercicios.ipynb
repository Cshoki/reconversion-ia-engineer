```python
#El más completo: tenés ventas_dia = [3200, 8900, 1500,
#6700, 4300, 9800, 2100]. Calculá, en el mismo loop:
#cuántas ventas superaron los 5000, y la suma total de
#esas ventas que superaron los 5000 (pista: acá el if
#filtra, y dentro de ese if van las dos actualizaciones,
#contador y total).

ventas_dia=[3200,8900,1500,6700,4300,9800,2100]
suma=0
contador=0

for ventas in ventas_dia:
  if ventas > 5000:
    contador+=1
    suma=suma+ventas
print("Ventas mayores a 5000: ", contador)
print("Suma ventas mayores a 5000: $", suma)
```

```python
#El más largo: tenés transacciones = [1500, -300, 4200,
#-800, 2100, -150, 3300] (los negativos representan
#gastos, los positivos ingresos). Calculá, en un mismo
#loop: el total de ingresos (solo positivos), el total
#de gastos (solo negativos, como número positivo —
#pista: podés sumar -numero cuando es negativo), y
#cuántas transacciones fueron gastos.

transacciones = [1500, -300, 4200,-800, 2100, -150, 3300]
suma_positivos=0
suma_negativos=0
contador_negativo=0

for transaccion in transacciones:
  if transaccion >0:
    suma_positivos+=transaccion
  else:
    suma_negativos+= abs(transaccion)
    contador_negativo+=1
print(suma_positivos)
print(suma_negativos)
print(contador_negativo)
```

```python
#Tenés stock = [12, 0, 8, 0, 5, 3, 0]. Contá cuántas
#posiciones tienen stock en 0 (esto es el patrón
#"contador" de siempre, no el de índice — para que
#compares los dos en la misma tanda).

stock = [12, 0, 8, 0, 5, 3, 0]
contador=0

for stock_1 in stock:
  if stock_1 ==0:
    contador+=1
print("En stock 0 hay: ", contador)
```

```python
#El que mezcla los dos patrones en un mismo loop:
#tenés pedidos = ["pendiente", "enviado", "pendiente",
#"entregado", "pendiente"]. Necesitás dos cosas a la .
#vez: la posición del primer "entregado" que aparezca,
#y cuántos "pendiente" hay en total en toda la lista.

pedidos = ["pendiente", "enviado", "pendiente", "entregado", "pendiente"]
contador=0
posicion=0

for pedido in pedidos:
  if pedido=="entregado":
    print("El primer entregado está en posición: ", posicion)
  posicion+=1
  if pedido=="pendiente":
    contador+=1
print(contador)
```

```python
#Tenés temperaturas = [22, 19, 31, 28, 17, 33, 20].
#Encontrá la posición de la primera temperatura mayor
#a 30.

temperaturas = [22, 19, 31, 28, 17, 33, 20]
posicion=0
ya_encontrado=False

for temperatura in temperaturas:
  if temperatura > 30 and ya_encontrado==False:
    print("La primera temperatura mayor a 30 es: ",posicion)
    ya_encontrado=True
  posicion+=1
```
